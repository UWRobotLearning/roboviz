```python
# change to app.ipynb
%pip install h5py
import git
import os
import subprocess

# Locate GitHub repo and clone it if not already done
repo_url = "https://github.com/UWRobotLearning/roboviz.git"  # Our roboviz repo URL
repo_dir = "current_code"  # Folder to clone the repo into

if not os.path.exists(repo_dir):
    print("Cloning the repository...")
    git.Repo.clone_from(repo_url, repo_dir)
else:
    print("Repository already cloned.")
    
# Pull the latest version of the repository (if already cloned)
repo = git.Repo(repo_dir)
origin = repo.remotes.origin
origin.pull()  # Fetches and merges the latest code

# Change "./path-to-whatever-file-we're-streaming-in-plotly"
script_path = os.path.join(repo_dir, "./roboviz/marco/marco_algorithm.py")

# Run the script
if os.path.exists(script_path):
    print("Running the script...")
    try:
        result = subprocess.run(
            ['python', script_path, "./expert_demos.hdf5"],
            capture_output = True,  # Check for errors
            text = True  # Ensure the output is in text format (not bytes)
        )
        # Print both stdout and stderr
        if result.stdout:
            print("Output:\n", result.stdout)
        
        if result.stderr:
            print("Error:\n", result.stderr)
    
    except Exception as e:
        print(f"An error occurred while running the script: {e}")
else:
    print(f"Script {script_path} not found.")
```
